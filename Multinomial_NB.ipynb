```python
import pandas as pd

# multinomial naive bayes import
from sklearn.naive_bayes import MultinomialNB

from sklearn.feature_extraction.text import CountVectorizer

#output : labels (spam/non-spam) --> spam=1 , non-spam=0

from sklearn.preprocessing import LabelEncoder

```

```python
# data
df = pd.read_csv("/content/dataset.csv")
df
```

```python
# input and output

x=df['email']#input
y=df['category']#output

model=MultinomialNB()

le=LabelEncoder()

#le.fit_transform --> means fit is for train(learn from past data) and transform convert text into number
y_val=le.fit_transform(y)

#convert y labels into values


#count vectorizer - to count each word in input(x)
vector=CountVectorizer()
x_val=vector.fit_transform(x)
#count words in x column



#model created

model.fit(x_val,y_val)
```

```python
#testing
while True:
  email_user=input("compose your mail here: ")

  test_email=vector.transform([email_user])


  prediction=model.predict(test_email)
  print("Predicted Email is :",le.inverse_transform(prediction)[0])

print("Predicted Email is :",prediction[0])

#spam - 1 and non-spam - 0
```
