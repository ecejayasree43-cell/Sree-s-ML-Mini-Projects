```python
# Exploratory Data Analysis (EDA)
# Objective

# EDA helps us understand the data visually before building the machine learning model.

# We'll use:

# 1 Univariate Plots
# 3 Bivariate Plots
# 1 Multivariate Plot
```

```python
# Univariate Analysis
# Plot 1: Histogram of Transaction Amount
# Purpose

# To understand how transaction amounts are distributed.

plt.figure(figsize=(8,5))

sns.histplot(df["Amount"], bins=50, kde=True)

plt.title("Distribution of Transaction Amount")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.show()

```

```python
#  Hist plot: Most transactions have small amounts, while only a few transactions have very high amounts.
```

```python
# Plot 2: Violin Plot - Amount Distribution by Class
# Why?

# Violin plot combines:

# Box plot information
# Distribution density

plt.figure(figsize=(8,5))

sns.violinplot(
    x="Class",
    y="Amount",
    data=df
)

plt.xticks(
    [0,1],
    ["Genuine","Fraud"]
)

plt.title("Transaction Amount Distribution by Fraud Class")

plt.show()
```

```python
# The violin plot shows "how transaction amounts are distributed across genuine and fraudulent classes."
# It helps identify whether "fraud transactions" have unique amount patterns.
```

```python
# Scatter Plot (Time vs Amount)
# Purpose

# To observe how transaction amounts vary over time.

```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))

sns.scatterplot(
    x="Time",
    y="Amount",
    data=df,
    alpha=0.5
)

plt.title("Transaction Amount over Time")
plt.xlabel("Time")
plt.ylabel("Amount")

plt.show()
```

```python
# Most transactions occur with lower amounts throughout the time period,
# while a few large transactions are scattered across the dataset.
```

```python
# Bar Chart (Average Amount by Class)
# Purpose

# To compare the average transaction amount between genuine and fraudulent transactions.
```

```python
plt.figure(figsize=(6,5))

sns.barplot(
    x="Class",
    y="Amount",
    data=df,
    estimator="mean"
)

plt.xticks([0,1],["Genuine","Fraud"])

plt.title("Average Transaction Amount by Class")
plt.xlabel("Transaction Type")
plt.ylabel("Average Amount")

plt.show()
```

```python
# logistic regression

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix

df=df.dropna()
# input feature
x=df[['V1', 'V2', 'V3', 'V4','Amount']]
y=df['Class']
model=LogisticRegression()
model.fit(x,y)
y_pred=model.predict(x)
print(accuracy_score(y,y_pred))
print(confusion_matrix(y,y_pred))

# user inputs to predict fraud or non-fraud
v1=float(input("Enter V1:"))
v2=float(input("Enter V2:"))
v3=float(input("Enter V3:"))
v4=float(input("Enter V4:"))
amount=float(input("Enter Amount:"))
new_data=pd.DataFrame({
    'V1':[v1],
    'V2':[v2],
    'V3':[v3],
    'V4':[v4],
    'Amount':[amount]
})

pred=model.predict(new_data)
print("Predicted Result:",pred[0])

if pred[0]==0:
    print("Genuine Transaction")
else:
    print("Fraudulent Transaction")

# df.columns
```
