```python
locations=df[['Latitude', 'Longitude']].copy()
```

```python
# convert coordinates to numeric
locations['Latitude']=pd.to_numeric(locations['Latitude'], errors='coerce')
locations['Longitude']=pd.to_numeric(locations['Longitude'], errors='coerce')
```

```python
# let remove missing and invalid coordinates
locations=locations.dropna(subset=['Latitude', 'Longitude'])
```

```python
locations=locations[locations['Latitude'].between(-90,90)& locations['Latitude'].between(-180,180)].copy()
```

```python
# remove coordinates that represents (0,0)
locations = locations[~((locations['Latitude'] == 0) & (locations['Longitude'] == 0))].copy()

print("Valid Location Records: ", len(locations))
```

```python
# convert latitude and longitude to radians
coords=np.radians(locations[['Latitude', 'Longitude']].to_numpy())
```

```python
# apply dbscan clustering
earth_radiuskm=6371.0
radiuskm=0.5

# eps radius --> Locations within roughly 500 metres are considered neighbours.

#minsamples=10 -->core point needs at least 10 points in its neighbourhood, including itself.

model=DBSCAN(eps=radiuskm/earth_radiuskm, min_samples=1, algorithm='ball_tree', metric='haversine')
locations['Cluster']=model.fit_predict(coords)
```

```python
# Display cluster results
cluster_ids = set(locations['Cluster'])

number_of_clusters = len(cluster_ids - {-1})
noise_count = (locations['Cluster'] == -1).sum()

print("Total clusters:", number_of_clusters)
print("Noise points:", noise_count)
print("\nCluster counts:")
print(locations['Cluster'].value_counts().sort_index())
```

```python
# Create the interactive Folium map
crime_map = folium.Map(location=[40.7128, -74.0060],zoom_start=11,tiles='CartoDB positron')
```

```python
#Assign a different colour to each cluster
colors = [
    'blue', 'red', 'green', 'purple', 'orange',
    'darkred', 'cadetblue', 'darkgreen', 'pink',
    'darkblue', 'beige', 'gray'
]
```

```python
 #Plot the arrest locations on the map
for _, row in locations.iterrows():

    cluster = int(row['Cluster'])

    if cluster == -1:
        color = 'black'
    else:
        color = colors[cluster % len(colors)]

    folium.CircleMarker(
        location=[
            row['Latitude'],
            row['Longitude']
        ],
        radius=3,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.65,
        weight=1,
        popup=f"Cluster: {cluster}"
    ).add_to(crime_map)
```

```python
#Save the interactive map
map_file = os.path.abspath("dbscan_crime_hotspots.html")
crime_map.save(map_file)

print("\nMap saved at:", map_file)
```
