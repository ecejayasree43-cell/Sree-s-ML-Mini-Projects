```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
#1. get website url

url="https://books.toscrape.com/"

#2. get webpage data
resp=requests.get(url)

# 3.check website response
print("Status Code:",resp.status_code)
```

```python
# 4.read webpage html
soup=BeautifulSoup(resp.text,"html.parser")

# 5.let extract first book title
book=soup.select_one("article.product_pod h3 a")

#  6.extract book price
price=soup.select_one("article.product_pod p.price_color")

print("Book Title:",book.text)
print("Book Price:",price.text)

# 7. extracct book availability
availability=soup.select_one("article.product_pod p.availability")
print("Book Availability:",availability.text.strip())
```

```python
# 8. store details in a dictionary
book_details={
     "Title":book['title'],
     "Price":price.text,
     "Availability":availability.text.strip()

}
```

```python
#9.Display details in a dictionary
print("Book Details:")
for key,value in book_details.items():
    print(f"{key}:{value}")
```

```python
# 10. store details in a dataframe
df=pd.DataFrame([book_details])
df
```

```python
# 11. Save details to csv format
csv_file=df.to_csv("webscrapebook.csv",index=False)
```

```python
excel_file=df.to_excel("webscrapebook.xlsx",index=False)
```
