```python
import pandas as pd

# BernoulliNB -- it simply check whether word present or not

#data will be take as one hot encoding --> if data is present , return 1
                                          #  otherwise return 0

from sklearn.naive_bayes import BernoulliNB

from sklearn.model_selection import train_test_split

# accuracy
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("/content/patient_churn_dataset.csv")
df
```

```python
# input and output features

df['missed']=(df['Missed_Appointments']>=0).astype(int)

df['day_visit']=(df['Days_Since_Last_Visit']>=180).astype(int)

# df['Days_Since_Last_Visit'].mean()
df['rating']=(df['Overall_Satisfaction']>=2).astype(int)

```

```python
x=df[['missed','day_visit','rating']]
y=df['Churned']

# input and output features
```

```python
# train and test split
x_train,x_test,y_train,y_test=train_test_split(x,y)
model=BernoulliNB()
model.fit(x_train,y_train)

```

```python
# accuracy
y_pred=model.predict(x_test)
acc=accuracy_score(y_test,y_pred)

print("Accuracy:",acc)
```

```python
# user inputs
# missed','day_visit','rating'

missed=int(input("Enter missed apointments:"))
day_visit=int(input("Enter day visits:"))
rating=int(input("Enter rating:"))

user=model.predict([[missed,day_visit,rating]])

if user==1:
  print("Churned")
else:
  print("Not Churned")
#  note:
#  (Binary form data will be taken and predict category)
# inputs: present or not? then it takes value as 0 or 1 and provide predicted result
```
