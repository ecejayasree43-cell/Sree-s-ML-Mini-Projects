```python
# Project objective:
# Analyze a collection of text,
#  identify frequently used words,
#   visualize important keywords,
#   and compare stemming with lemmatization.
```

```python
import nltk
import re
import matplotlib.pyplot as plt

from collections import Counter
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from wordcloud import WordCloud

nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# User input
text = input("Enter your text for NLP analysis: ")

# Text cleaning
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

stop_words = set(stopwords.words('english'))
clean_words = [w for w in words if w not in stop_words]
```

```python

# Text mining
frequency = Counter(clean_words)
print("\nWord Frequency:", frequency.most_common(10))

# Word cloud
if clean_words:
    cloud = WordCloud(background_color="white").generate_from_frequencies(frequency)
    plt.imshow(cloud, interpolation="bilinear")
    plt.axis("off")
    plt.show()
```

```python
# Stemming and lemmatization
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

print("\nWord Analysis:")
for word in clean_words:
    print(word, "-> Stem:", stemmer.stem(word),
          "| Lemma:", lemmatizer.lemmatize(word, pos="v"))
```
