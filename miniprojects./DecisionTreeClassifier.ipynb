```python
import pandas as pd

# import
from sklearn.tree import DecisionTreeClassifier

# output category

from sklearn.preprocessing import LabelEncoder

#train and test split

from sklearn.model_selection import train_test_split

#accuracy score

from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("/content/drug200.csv")
df
```

```python
# input and output features



le_sex=LabelEncoder()
le_bp=LabelEncoder()
le_chol=LabelEncoder()


# convert labels into number and also fit into the model
df['sex']=le_sex.fit_transform(df['Sex'])
df['bp']=le_bp.fit_transform(df['BP'])
df['chol']=le_chol.fit_transform(df['Cholesterol'])

# df.columns
x=df[['sex','bp','chol','Age']]
y=df['Drug']
```

```python
# model create
model=DecisionTreeClassifier()

# train and test data

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
model.fit(x_train,y_train)
```

```python
# accuracy

y_pred=model.predict(x_test)

acc=accuracy_score(y_test,y_pred)

print("Accuracy: ",acc*100)
```

```python
# user input

# sex,bp,chol
age=int(input("Enter your Age:"))
sex=input("Enter Gender category(F or M):").upper()
bp=input("Enter BP Level(High or Low)").upper()
chol=input("Enter Cholestrol level(High or Normal)").upper()


# prediction
sex=le_sex.transform([sex])
bp=le_bp.transform([bp])
chol=le_chol.transform([chol])

# output




pred=model.predict([[sex[0],bp[0],chol[0],age]])[0]
print("Predicted Result:",pred)
```
