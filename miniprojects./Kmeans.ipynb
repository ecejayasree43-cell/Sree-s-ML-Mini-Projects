```python
# problem  statement: identify poor,avg and top category students



sub=['math_score', 'history_score', 'physics_score',
       'chemistry_score', 'biology_score', 'english_score', 'geography_score']
df['avg_score']=df[sub].mean(axis=1)
```

```python
df.head()
```

```python
df.columns
```

```python
x=df[['avg_score','absence_days','weekly_self_study_hours']]
```

```python
model=KMeans(n_clusters=3,random_state=42)
```

```python
model.fit(x)
```

```python
# predict

math=int(input("Enter Maths score:"))
history=int(input("Enter History score:"))
physics=int(input("Enter Physics score:"))
chemistry=int(input("Enter Chemistry score:"))
biology=int(input("Enter Biology score:"))
english=int(input("Enter English score:"))
geography=int(input("Enter Geography score:"))
studyhours=int(input("Enter Study hours:"))
absence=int(input("Enter Absence days:"))

avg=math+history+physics+chemistry+biology+english+geography/7


pred=model.predict([[avg,studyhours,absence]])[0]
print("Predicted category: ",pred)

```

```python
if pred==0:
  print("Poor Category")
elif pred==1:
  print("Average Category")
else:
  print("Topper category")
```
