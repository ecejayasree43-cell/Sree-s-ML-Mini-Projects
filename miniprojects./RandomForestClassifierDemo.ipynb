```python
import pandas as pd

# Random Forest Classifier
from sklearn.ensemble import RandomForestClassifier

# Label Encoder
from sklearn.preprocessing import LabelEncoder

#test and train split
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("/content/tested.csv")
df
```

```python
df.head()
```

```python
df.tail()
```

```python
df.isnull().sum()
```

```python
# delete cabin column
df.drop('Cabin',axis=1,inplace=True)
```

```python
df.head()
```

```python
df.columns
# new feature engineering
df['FamilySize']=df['SibSp']+df['Parch']+1
df.head()
```

```python
# sex and Embarked feature

```

```python
le_emd=LabelEncoder()
le_sex=LabelEncoder()
```

```python
df['Sex']=le_sex.fit_transform(df['Sex'])
df['Embarked']=le_emd.fit_transform(df['Embarked'])
```

```python
# input and output feature

x=df[['Sex','Age','Embarked','FamilySize']]
y=df['Survived']
```

```python
model=RandomForestClassifier()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

```

```python
model.fit(x_train,y_train)
```

```python
# y_pred

y_pred=model.predict(x_test)
```

```python
# accuracy score
from sklearn.metrics import accuracy_score

acc=accuracy_score(y_test,y_pred)
print("Accuracy: ",acc)
```

```python
# user input

name=input("Enter your name:")
age=int(input("Enter your age:"))
gen=input("Enter your Gender:").strip().lower()
emb=input("Enter Your Embarkment (Q | C | S)").strip().upper()
family=int(input("Enter Familysize"))


gen=le_sex.transform([gen])[0]
emb=le_emd.transform([emb])[0]

pred=model.predict([[age,gen,emb,family]])

if pred[0]==1:
  print("Survived")
else:
  print("Not Survived")
```
