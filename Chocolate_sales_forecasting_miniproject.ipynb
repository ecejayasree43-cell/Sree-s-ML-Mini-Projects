```python
df['Date']=pd.to_datetime(df['date'])
df['Date']
```

```python
df=df.sort_values("date")
df
```

```python
data=df['day_of_week']
# df.head()
```

```python
model=ARIMA(data,order=[6,1,0])
model.fit=model.fit()

pred=model.fit.forecast(steps=7)
```

```python
print("Forecasting result for chocolate Sales:")
print("Day 1:",pred.iloc[0])
print("Day 2:",pred.iloc[1])
print("Day 3:",pred.iloc[2])
print("Day 4:",pred.iloc[3])
print("Day 5:",pred.iloc[4])
print("Day 6:",pred.iloc[5])
print("Day 7:",pred.iloc[6])
```
