```python
# let grp into 3 clusters
# 1. High performance- Topper
# 2. Average Performance - Average
# 3. Below Average Performance - Poor or Slow Learner


# x=> features like avg score= sum of ('math_score', 'history_score', 'physics_score',
#        'chemistry_score', 'biology_score', 'english_score', 'geography_score')
# y=> feature like  'weekly_self_study_hours'
```

```python
subjects=['math_score', 'history_score', 'physics_score',
       'chemistry_score', 'biology_score', 'english_score', 'geography_score']

df['avg_score']=df[subjects].mean(axis=1)
```

```python
x=df[['avg_score','weekly_self_study_hours']]

```

```python
model=KMeans(n_clusters=3,random_state=34)
```

```python
model.fit(x)
```

```python
labels=model.labels_
centroid=model.cluster_centers_
```

```python
 # let get input from user to find avg_score for new student 
math = float(input("Enter Math Marks: "))
history = float(input("Enter History Marks: "))
physics = float(input("Enter Physics Marks: "))
chemistry = float(input("Enter Chemistry Marks: "))
biology = float(input("Enter Biology Marks: "))
english = float(input("Enter English Marks: "))
geography = float(input("Enter Geography Marks: "))

study_hours = float(input("Enter Weekly Self Study Hours: "))
```

```python
average = (
    math +
    history +
    physics +
    chemistry +
    biology +
    english +
    geography
) / 7
```

```python
# prediction of performance of new student
newstudent=[[average,study_hours]]
clusters=model.predict(newstudent)
```

```python
print("Prediction Result: ",clusters[0])
```

```python
if clusters[0]==0:
    print("Category :Poor Performance! Need Improvement!")
elif clusters[0]==1:
    print("Category : Average Performance!")
else:
    print("Category : Excellent Performance")
```
