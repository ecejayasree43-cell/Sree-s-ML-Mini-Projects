```python
import pandas as pd
from sklearn.naive_bayes import BernoulliNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("/content/patient_churn_dataset.csv")
df
```

```python
#let take numeric inputs
#for bernoulli NB - inputs can be numeric or text --> main thing that particular word or number present or not
df['Missed'] = (df['Missed_Appointments'] > 0).astype(int)

df['Long_Gap'] = (df['Days_Since_Last_Visit'] >= 180).astype(int)

df['Low_Satisfaction'] = (df['Overall_Satisfaction'] <= 2).astype(int)

```

```python
# df.head()

x=df[['Missed','Long_Gap','Low_Satisfaction']]
y=df['Churned']
df.columns
```

```python
model=BernoulliNB()


x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model.fit(x_train,y_train)

le=LabelEncoder()

y_pred=model.predict(x_test)

acc=accuracy_score(y_test,y_pred)

print("Accuracy: ",acc*100,"%")
```

```python
le.fit(["No", "Yes"])

missed = input("Is appointment missed? (Yes/No): ")
Long_Gap = input("Is there a long gap? (Yes/No): ")
Low_Satisfaction = input("Is satisfaction low? (Yes/No): ")

# Convert Yes/No → 0/1
missed = le.fit_transform([missed])[0]
Long_Gap = le.fit_transform([Long_Gap])[0]
Low_Satisfaction = le.fit_transform([Low_Satisfaction])[0]

prediction=model.predict([[missed,Long_Gap,Low_Satisfaction]])
print("Predicted Result",le.fit_transform(prediction)[0])

if prediction[0] == 1:
    print("Predicted Result: Patient will CHURN")
else:
    print("Predicted Result: Patient will NOT CHURN")
```
